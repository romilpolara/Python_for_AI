### Sequence sum

```python
n = int(input("Enter the the value of n: "))
result = 0
fect = 1

for i  in range(1, n + 1):
    fect =  fect * i
    result = result + i/fect

print("The sum of factorials from 1 to", n, "is:", result)
```

### Break
break is used when you want to completely terminate the loop.
 
EXample - Liner Search

```python
num1 = int(input("Enter the value of num1: "))
num2 = int(input("Enter the value of numm2: "))
for i in range(num1, num2 + 1):
    for j in range(2, i):
        if i%j == 0:
            break
    else:
        print(i)
```

# Python String — Logical Operations

A string is a sequence of characters.

Python allows us to perform several logical operations on strings using operators such as:

- `==` → Equal
- `!=` → Not Equal
- `in` → Membership
- `not in` → Not Membership
- `and` → Logical AND
- `or` → Logical OR
- `not` → Logical NOT
- `is` → Object Identity

These operations are important for Python placement and interview preparation.

# 1. String Comparison

Python allows us to compare two strings.

## `==` — Equal

Checks whether two strings contain the same characters.

### Example

```python
s1 = "hello"
s2 = "hello"

print(s1 == s2)
```

## `!=` — Not Equal

Checks whether two strings contain different values.

```python
s1 = "hello"
s2 = "world"

print(s1 != s2)
```

## Other Comparison Operators

We can also compare strings using:

- `<` → Less than
- `>` → Greater than
- `<=` → Less than or equal to
- `>=` → Greater than or equal to

### Example

Python compares strings **lexicographically**, roughly similar to dictionary ordering.

```python
print("apple" < "banana")
```

### Output

`True`

Because `"apple"` comes before `"banana"` in lexicographical order.

# 2. How String Comparison Works

Python compares strings character by character from **left to right**.

For example:

    apple
    banana
    ↑
    a comes before b

Therefore:

    "apple" < "banana"

returns `True`.


### Important

String comparison is **case-sensitive**.

For example:

    "Python" != "python"

because uppercase `P` and lowercase `p` are different characters.

# 3. `in` Operator

The `in` operator checks whether a character or substring exists inside a string.

### Syntax

    substring in string

It returns either `True` or `False`.

```python
s = 'hello world'

print('python' in s)
print('z' not in s) # The `not in` operator checks whether a character or substring **does not exist** inside a string.

### Placement Example

# We can check whether an email contains the `@` symbol.
```

# 5. `and` Operator

The `and` operator is used when **multiple conditions must be true**.

' '(blank) = False

'string'(any string) = True

### Truth Table

| Condition 1 | Condition 2 | Result |
|-------------|-------------|--------|
| True        | True        | True   |
| True        | False       | False  |
| False       | True        | False  |
| False       | False       | False  |

### Example

Both username and password must be correct.

# 6. `or` Operator

The `or` operator returns a truthy result when **at least one condition is true**.

### Truth Table

| Condition 1 | Condition 2 | Result |
|-------------|-------------|--------|
| True        | True        | True   |
| True        | False       | True   |
| False       | True        | True   |
| False       | False       | False  |

### Real-Life Example

Saturday OR Sunday is a weekend.

# 7. `not` Operator

The `not` operator reverses a Boolean result.

    True  → False
    False → True

```python
s = "Python"

print("Java" in s)
print(not ("Java" in s))
```

# 9. Logical Operators Directly on Strings ⭐⭐⭐

A very important Python interview concept:

`and` and `or` do not always return `True` or `False`.

They can return one of their operands.

## Example

```python
print("Hello" and "World")
```

### Output

    World

Because `"Hello"` is a non-empty string and therefore truthy.

Now consider:

```python
print("Hello" or "World")
```

### Output

    Hello

This happens because `"Hello"` is already truthy.

This behavior is related to **short-circuit evaluation**.

# 10. Understanding `and`

For:

    A and B

Python evaluates `A` first.

- If `A` is falsy → return `A`
- If `A` is truthy → evaluate and return `B`

### Example 1

```python
print("" and "Hello")
```

### Output

    ""

Because the empty string is falsy.

### Example 2

```python
print("Python" and "Java")
```

### Output

    Java

Because `"Python"` is truthy, Python evaluates the second operand and returns `"Java"`.

# 11. Understanding `or`

For:

    A or B

Python evaluates `A` first.

- If `A` is truthy → return `A`
- If `A` is falsy → evaluate and return `B`

### Example 1

```python
print("Python" or "Java")
```

### Output

    Python

Because `"Python"` is already truthy.

### Example 2

```python
print("" or "Java")
```

### Output

    Java

Because `""` is falsy, Python evaluates the second operand.
