# `if-else` — ATM Machine 💳

An ATM needs to decide whether the user can withdraw money.

```python
balance = 10000
withdraw = 5000

if withdraw <= balance:
    balance = balance - withdraw
    print("Transaction successful")
    print("Remaining balance:", balance)
else:
    print("Insufficient balance")
```

## Real-Life Logic

```text

       ↓
Is amount <= balance?
    /          \
  YES           NO
   ↓             ↓
Withdraw      Reject
money         transaction
```

# `while` loop — Elevator System 🛗

An elevator can keep moving toward a target floor **while it has not reached that floor**.

```python
current_floor = 1
target_floor = 5

while current_floor < target_floor:
    current_floor += 1
    print("Elevator is on floor", current_floor)

print("Target floor reached")
```

## Output

```text
Elevator is on floor 2
Elevator is on floor 3
Elevator is on floor 4
Elevator is on floor 5
Target floor reached
```

## Real-Life Logic

```text
Current floor = 1
       ↓
Is current floor < target floor?
       ↓
      YES
       ↓
Move up one floor
       ↓
Check condition again
       ↓
       ...
       ↓
Current floor = 5
       ↓
      NO
       ↓
Stop
```

## Placement Explanation

> "`while` loop is used when we want to repeatedly execute a block of code as long as a particular condition remains true."


````python
#Guessing game

import random
jackpot = random.randint(1, 100) 

guess = int(input("guess a number between 1 to 100: ")) 
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print(f"{guess} is lower than the jackpot.")
    else:
        print(f"{guess} is higher   than the jackpot.")
    
    guess = int(input("guess a number between 1 to 100: "))
    counter += 1
else: 
    print("correct guess", jackpot)
    print("counter is: ", counter )
````

# `for` loop — Shopping Cart 🛒

Suppose an online shopping website needs to calculate the **total price of all products** in a customer's cart.

```python
prices = [500, 1200, 300, 800]

total = 0

for price in prices:
    total = total + price

print("Total price:", total)
```

### Output

```text
Total price: 2800
```

### Real-Life Logic

```text
Shopping Cart
      ↓
Take first product price
      ↓
Add price to total
      ↓
Take next product price
      ↓
Add price to total
      ↓
Continue for every product
      ↓
All products processed
      ↓
Display total price
```

### Placement Explanation

> "`for` loop is used when we need to repeat an operation for each item in a collection, such as calculating the total price of products in a shopping cart."


````python
# for loop exmple

total_population = 10000

for i in range(1,11):
    print("Year: ", i, "Total population is: ", total_population)
    total_population += (total_population)* 10/100

````

````python
total_population = 10000

for i in range(10, 0, -1):
    print("Year: ", i, "Total population is: ", total_population)
    total_population -= (total_population)* 10/100
````
