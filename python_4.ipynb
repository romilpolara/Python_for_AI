## What is a List?

A **list** is a built-in Python data structure used to store multiple values in a single variable.

A list:

- **is ordered** — maintains the order of elements.
- **is mutable** — elements can be changed after the list is created.
- **allows duplicate values** — the same value can appear multiple times.
- **can store different data types** — such as `int`, `float`, `str`, and `bool`.
- **supports indexing** — elements can be accessed using their index.
- **supports slicing** — a portion of the list can be extracted.
- **can contain other lists** — this is called a **nested list**.

## List Comprehension

List comprehension is provides a concise way of creating lists.

newlist = [expression for item in iterable if condition == True]

#### Advantages of List Comprehension
- More time-efficient and space-efficient than loops.
- Requires fewer lines of code.
- Transforms an iterative statement into a formula.

````python
# Add 1 to 10 numvers to a list
list1 = [i for i in  range(1, 11)]
print(list1)
````

````python
# Scalar multiplication on a vector
v = [2,3,5]
s = -3
list1 = [s * i for i in v]
print(list1)
````

````python
# Add Squares
l1 = [2,4,5,6,7,3]

list1 = [i ** 2 for i in l1]
print(list1)
````

````python
# print all numbers divisible by 5 in the range of 1 to 50

list1 = [i for i in range(1, 51) if i%5==0]
print(list1)
````

````python
# find language start with letter p

languages = ['java', 'python', 'php', 'c', 'javascript']

list1 = [i for i in languages if i[0] == 'p']
print(list1)
````

````python
# Nested if with list comprehension
basket = ['apple', 'guava', 'cherry', 'banana']
my_fruits = ['apple', 'kiwi', 'grapes', 'banana']

my_fruits = [i for i in my_fruits if i in basket and i.startswith('a')]

print(my_fruits)
````

````python
# print a (3,3) matrix using list comprehension -> nested list comprehension

list1 = [[i * j for i in range(1, 4)] for j in range(1,4)]
print(list1)
````

````python
# Cartesion products
l1 = [1,2,3,4]
l2 = [5,6,7,8]

list1 = [i * j for i in l1 for j in l2]
print(list1)
````

#### 2 ways to traverse a list

````python
# itemwise
l = [1,2,3,4]

for i in l:
    print(i)
````

````python
# indexwise 
l = [1,2,3,4]

for i in range(0, len(l)):
    print(l[i])
````
