```python
# Find the frequency

s = input("Enter the string: ")
print(s)
term = input("What would like to search for: ")
print(term)
counter = 0
for i in s:
    if i == term:
        counter +=1

print('Frequency:', counter)
```

```python
# Write a program which can remove a particular character form a string

s = input("Enter the string: ")
print(s)
term = input("What would like to search for: ")
print(term)

list1= []
for i in s:
    if i == term:
        continue
    else:
        list1.append(i)

print("".join(list1))
```

```python
# Write a program that can check whether a given string  is palindrome or not.

s = input("Enter the string: ")
print(s)
if s == s[::-1]:
    print("This is a palindrome stirng.")

else: 
    print("This is not a palindrome string.")

```

```python
#Write a program to count the number of words in a string without  split()

s = input("Enter the string: ")
print(s)

list1 = []
temp = ''
for i in s:
    if i != " ":
        temp+=i
    else: 
        list1.append(temp)
        temp = ''

list1.append(temp)
print(list1)
print((len(list1)))
```

```python
# Write a python program to convert a string to titlecase wihtout using the title().

s = input("Enter the string: ")
print(s)

list1 = []
for i in s.split():
    list1.append(i[0].upper() + i[1:].lower())

print("Output is:"," ".join(list1))
```

```python
# write a program that can convert an integer to string.
s = int(input("Enter the integer value: "))
print(s)
print(type(s))

digit = '0123456789'
temp = ''

while s != 0:
    temp = digit[s%10] + temp
    s = s//10

print(temp)
print(type(temp))
```
